```python
import pandas as pd
```

```python
data = pd.read_csv('data/bitcoin2.csv')
data
```

```python
import re
def cleaning(sent):
    #Mengubah kata menjadi huruf kecil semua dengan fungsi lower()
    string = sent.lower()
    string = re.sub(r'http\S+', '', string) #Menghapus URL
    string = re.sub(r'@\w+', '', string) #Menghapus mention
    string = re.sub(r'#\w+', '', string) #Menghapus hashtag
    string = re.sub(r'\d+', '', string) #Menghapus angka
    string = re.sub(r'[^\w\s]', '', string) #Menghapus karakter khusus
    return string
```

```python
data['full_text'] = data.full_text.apply(cleaning)
data.head()
```

```python
data = data[['full_text']]
data.head()
```

```python
data.to_csv('data/bitcoin2_clean.csv', index=False)
```
